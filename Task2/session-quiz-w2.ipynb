***Qestion 2.12 - session quiz***

```python
from math import atan2, degrees
x = float(input("Enter a number for the value of x:"))
y = float(input("Enter a number for the value of y:"))
angl=degrees(atan2(y,x))
if -90>angl:
    print(f"The angle between the positive x-axis and the line l((0,0),P(x,y)) is {angl} and P is in the Third quadrant.")
elif 0>angl:
    print(f"The angle between the positive x-axis and the line l((0,0),P(x,y)) is {angl} and P is in the Fourth quadrant.")
elif 90>angl:
    print(f"The angle between the positive x-axis and the line l((0,0),P(x,y)) is {angl} and P is in the First quadrant.")
elif 180>=angl:
    print(f"The angle between the positive x-axis and the line l((0,0),P(x,y)) is {angl} and P is in the Second quadrant.")
else:
    print("Malfunction detected, the angle between the positive x-axis and the line l((0,0),P(x,y)) is {angl}")
```

***Question 2.15 - session quiz***

```python
night1 = [1,3,2,4,0,3,7,5,2,9,2,4,3]
night2 = [0,0,2,4,2,4,3,2,1,1,1,2,0]
night1 = set(night1)
night2 = set(night2)
inters = set.intersection(night1, night2)
print(inters)
if 5 in inters:
    print("The element 5 is in the intersection.")
else:
    print("The element 5 is not in the intersection.")
if 5 in night1:
    print("Since '5 in night1' is True, 5 is part of night1")
else:
    print("Since '5 in night1' is False, 5 is not part of night1")
```
