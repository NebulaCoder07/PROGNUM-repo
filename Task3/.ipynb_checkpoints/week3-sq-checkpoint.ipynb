**Question 3.1**

```python
masses = [1.9891e+30, 1.8986e+27, 
          5.6846e+26, 1.0243e+26, 8.6810e+25,
          5.9736e+24, 4.8685e+24, 6.4185e+23, 
          3.3022e+23, 7.349e+22, 1.25e+22]
names = ["Sun", "Jupiter", "Saturn", "Neptune", "Uranus", "Earth", "Venus", "Mars", "Mercury", "Moon", "Pluto"]
smaller_than_moon=[]
for mass in masses:
    if mass <= masses[names.index("Moon")]:
        smaller_than_moon.append(mass)

print(f"Masses that are <= than the Moon's mass: {smaller_than_moon}")
lastmass = masses[slice(len(masses)-5,len(masses))]
print(f"The last 5 masses are: {lastmass}")
print(f"The verage mass in the list is: {sum(lastmass)/len(lastmass)} kg.")
```

**Question 3.2**

```python
from scipy.constants import G
import math
masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22]
radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195]
names  = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto']
esc = []
full_data = {}

for i in range(len(masses)):
    esc.append(math.sqrt(2*G*masses[i]/(radii[i]*1000)))
    print(f"{names[i]}'s escape velocity is {esc[-1]/1000:.2f} km/s with mass {masses[i]: .2e} kg and radius {radii[i]} km.")
    full_data[names[i]] = esc[-1]
print()
print(full_data)
```

**Question 3.5**

```python
sum_range = [100,200,300,400,500,600]
sums = []
for r in sum_range:
    sums.append(sum(1/i**2 for i in range(1,r+1)))
conv = True
for i in range(len(sums)-2):
    if sums[i+2]-sums[i+1] > sums[i+1]-sums[i]:
        conv = False
        break
if conv:
    print("The function 1/x^2 converges.")
else:
    print("The function 1/x^2 does not converge.")
```

**Question 3.10**

```python
from math import pi
def cylinder(r,h):
    A = 2*pi*r*(r+h)
    V = pi*r**2*h
    return A,V
area, volume = cylinder(1,1)
print(f"The area:volume ratio of a cylinder with r=1 and h=1 is {area/volume}")
```

**Question 3.17**

```python
def stdev(array):
    #------MEAN------
    mean_of_elements = sum(array)/len(array)
    
    #------STDEV------
    import math
    dev_squared = 0
    for i in array:
        dev_squared += (i-mean_of_elements)**2/len(array) #calculates the square of the standard deviation
    dev = math.sqrt(dev_squared) #calculates the actual standard deviation

    return dev
test = np.random.rand(10)
print(f"The standard devation given by stdev() is {stdev(test)}, by np.std() is {np.std(test)}, their difference is {np.std(test)-stdev(test)}.")
```

```python
import math
def stdev(array):
    return math.sqrt(sum((i-sum(array)/len(array))**2/len(array) for i in array))
    
test = np.random.rand(10)
print(f"The standard devation given by stdev() is {stdev(test)}, by np.std() is {np.std(test)}, their difference is {np.std(test)-stdev(test)}.")
```

**Question 3.18**

```python
data = [18.49, 2094.75,
35.48, 2320.87,
35.48, 1224.9,
11.34, 872.13,
7.12, 432.2,
9.82,  751,
23.66, 1410,
44,    2700,
66.7,  5292,
52.4,  2550,
55,    5253,
25.6,  2419,
26.19, 2158]
distance = np.array(data[::2])
velocity = np.array(data[1::2])

mean_d = np.mean(distance)
mean_velo = np.mean(velocity)

b = sum((distance[i]-mean_d)*(velocity[i]-mean_velo) for i in range(len(distance)))/sum((distance[i]-mean_d)**2 for i in range(len(distance)))
a = mean_velo-b*mean_d

print(f"The best linear fit (a+bx) for the data has the parameters a={a:.2f} km/s/Mpc and b={b:.2f} km/s")
```

* The best linear fit $(a+bx)$ for the data has the parameters $a = 9.19\,km/s/Mpc$ and $b = 71.38\,km/s$.

**Question 3.22**

```python
import matplotlib.pyplot as plt
from math import cosh
x = range(-5,6)
y = [cosh(i) for i in x]
plt.title("Chain model made with range")
plt.ylabel("y = Altitude [m]", fontsize = 12)
plt.xlabel("x = Position [m]", fontsize = 12)
plt.grid()
plt.plot(x, y, label="$y = cosh(x)$", color='red', marker = 'o')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = np.arange(-5,6,1)
y = np.cosh(x)
plt.title("Chain model made with NumPy")
plt.ylabel("y = Altitude [m]", fontsize = 12)
plt.xlabel("x = Position [m]", fontsize = 12)
plt.grid()
plt.plot(x, y, label="$y = cosh(x)$", color='green', marker = 'o')
plt.legend()
plt.show()
```
