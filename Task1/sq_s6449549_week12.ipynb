**Question 1.19**

* '*' Means 'any' character
* 'ls *[0-9]*.[co]' gives the result: **t1.c  t2.o  t7b3.c**
* '[' and ']' mean one of them
* the second would only show results with '.co' extension, while the first one showed results with .c or .o extension
